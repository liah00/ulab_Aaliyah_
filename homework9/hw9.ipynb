# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python
def plotting_functions():
    # Define the domain of x
    x = np.linspace(0, 10, 10)      
    fx = x**2  
    gx = x**(1/2) 

    plt.figure(figsize=(10, 8))
    plt.plot(x, fx, label='f(x) = x^2', color='blue', linestyle='-') 
    plt.plot(x, gx, label='g(x) = x^(1/2)', color='red', linestyle='--')
        
    plt.xlabel('x - axis', fontsize=12)
    plt.ylabel('y -axis', fontsize=12)
    plt.title('square and squareroot', fontsize=12)
    plt.legend()

    plt.show()
    return list(fx), list(gx)

plotting_functions()
```

```python
def plotting_functions():
    # graphs x**2. x**(1/2), and their sum 
    x = np.linspace(0, 10, 10)      
    fx = x**2  
    gx = x**(1/2) 
    sum = fx + gx

    plt.figure(figsize=(10, 8))
    plt.plot(x, fx, label='f(x) = x^2', color='blue', linestyle='-') 
    plt.plot(x, gx, label='g(x) = x^(1/2)', color='red', linestyle='--')
    plt.plot(x, sum, label = 'sum', color= 'green', linestyle='-.')

    plt.axvline(x=5, color='orange', linestyle=':')
    plt.axhline(y=10, color='orange', linestyle=':')

    plt.xlim(0,20)
    plt.ylim(0,20)
    
    plt.xlabel('x - axis', fontsize=12)
    plt.ylabel('y -axis', fontsize=12)
    plt.title('square and squareroot', fontsize=12)
    plt.legend(bbox_to_anchor = (1.4,1))

    plt.show()
    return list(fx), list(gx)

plotting_functions()
```
